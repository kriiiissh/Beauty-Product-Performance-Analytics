# Sephora E-Commerce EDA: 02 - Product Catalog Analysis

This notebook analyzes Sephora's product catalog. We examine category hierarchy, price distribution (to find luxury pricing tiers), top brands, loves count, and flags like exclusives.

```python
import sys
import os
sys.path.append(os.path.abspath(".."))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_products
from src.viz_utils import set_custom_style, save_figure

set_custom_style()
products_df = load_products("../data/product_info.csv")
```

## 1. Category Breakdown

Let's see what primary categories make up Sephora's inventory.

```python
plt.figure(figsize=(10, 6))
cat_counts = products_df["primary_category"].value_counts()
sns.barplot(x=cat_counts.values, y=cat_counts.index, palette="magma")
plt.title("Product Count by Primary Category", fontsize=14, weight="bold")
plt.xlabel("Count")
plt.ylabel("Primary Category")
plt.show()
```

## 2. Price Distribution by Category

We look at prices by category. We limit our view to products <= $150 to remove extreme pricing outliers (like large sets) for better visual scale.

```python
plt.figure(figsize=(12, 6))
filtered_price_df = products_df[products_df["price_usd"] <= 150]
sns.boxplot(data=filtered_price_df, x="price_usd", y="primary_category", palette="crest")
plt.title("Price Distribution by Category (Products <= $150)", fontsize=14, weight="bold")
plt.xlabel("Price (USD)")
plt.ylabel("Category")
plt.show()
```

## 3. Brand Representation

Let's see who are the top 15 brands by product counts.

```python
top_brands = products_df["brand_name"].value_counts().head(15).reset_index()
top_brands.columns = ["brand_name", "count"]

plt.figure(figsize=(12, 6))
sns.barplot(data=top_brands, x="count", y="brand_name", palette="flare")
plt.title("Top 15 Brands by Product Count in Sephora Catalog", fontsize=14, weight="bold")
plt.xlabel("Number of Products")
plt.ylabel("Brand Name")
plt.show()
```

## 4. Correlation Matrix of Numeric Attributes

```python
plt.figure(figsize=(10, 8))
numeric_cols = ["loves_count", "rating", "reviews", "price_usd", "child_count"]
sns.heatmap(products_df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", square=True)
plt.title("Correlation Matrix of Product Catalog Attributes", fontsize=14, weight="bold")
plt.show()
```
